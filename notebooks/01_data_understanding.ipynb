```python
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent
RAW_DATA_DIR = PROJECT_ROOT / "data" / "raw"

FREQUENCY_FILE = RAW_DATA_DIR / "freMTPL2freq.pq"
SEVERITY_FILE = RAW_DATA_DIR / "freMTPL2sev.pq"

print("Proje ana klasörü:", PROJECT_ROOT)
print("Frequency dosyası:", FREQUENCY_FILE)
print("Severity dosyası:", SEVERITY_FILE)
```

```python
print("Frequency mevcut mu?", FREQUENCY_FILE.exists())
print("Severity mevcut mu?", SEVERITY_FILE.exists())
```
